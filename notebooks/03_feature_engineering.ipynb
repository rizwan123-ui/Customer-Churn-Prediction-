```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/processed/customer_churn_cleaned.csv")

print("Dataset Shape:", df.shape)
```

```python
# Convert required columns to numeric first
df["Total Charges"] = pd.to_numeric(df["Total Charges"], errors="coerce")
df["Tenure Months"] = pd.to_numeric(df["Tenure Months"], errors="coerce")

# Service Count
service_columns = [
    "Phone Service",
    "Online Security",
    "Online Backup",
    "Device Protection",
    "Tech Support",
    "Streaming TV",
    "Streaming Movies"
]

df["Service_Count"] = df[service_columns].apply(
    lambda row: sum(value == "Yes" for value in row),
    axis=1
)

# Average Monthly Spend
df["Average_Monthly_Spend"] = (
    df["Total Charges"] /
    df["Tenure Months"].replace(0, 1)
)

# Customer Tenure Group
df["Tenure_Group"] = pd.cut(
    df["Tenure Months"],
    bins=[-1, 12, 24, 48, 60, 100],
    labels=["0-12", "13-24", "25-48", "49-60", "61+"]
)

print(df[
    ["Service_Count", "Average_Monthly_Spend", "Tenure_Group"]
].head())
```

```python
df.to_csv(
    "../data/processed/customer_churn_features.csv",
    index=False
)

print("Feature engineered dataset saved!")
print("Shape:", df.shape)
```
